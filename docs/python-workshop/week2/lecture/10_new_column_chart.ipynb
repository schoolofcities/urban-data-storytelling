# Creating a new column and a simple chart

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
cities = cities.rename(
    columns={
        "Name": "city",
        "Prov/terr": "province",
        "Population, 2021": "population_2021",
        "Population, 2016": "population_2016"
    }
)
cities = cities.dropna(
    subset=["population_2021", "population_2016"]
)
```

Create a new column by assigning a calculation to a new column name.

```python
cities["population_change"] = (
    cities["population_2021"]
    - cities["population_2016"]
)

cities[["city", "population_change"]].head()
```

Prepare the data before plotting. Here we sort the municipalities and retain the ten largest.

```python
largest_cities = cities.sort_values(
    by="population_2021",
    ascending=False
).head(10)

largest_cities[["city", "population_2021"]]
```

```python
largest_cities.plot.bar(
    x="city",
    y="population_2021",
    title="Population of Canada’s ten largest municipalities",
    legend=False
)
```

The chart is exploratory: it helps us check and communicate a simple comparison. More detailed chart design will come later.
