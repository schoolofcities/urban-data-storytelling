# Renaming columns and identifying missing values

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
```

Rename columns when their original names are awkward to type or interpret.

```python
cities = cities.rename(
    columns={
        "Name": "city",
        "Prov/terr": "province",
        "Population, 2021": "population_2021",
        "Population, 2016": "population_2016"
    }
)

cities.head()
```

Missing values commonly appear as `NaN`. Count them before deciding what to do.

```python
cities.isna().sum()
```

Inspect rows with a missing 2016 population.

```python
cities.loc[cities["population_2016"].isna()]
```

Remove only rows missing values required for the planned calculation.

```python
cities_clean = cities.dropna(
    subset=["population_2021", "population_2016"]
)

cities_clean.shape
```
