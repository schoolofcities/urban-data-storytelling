# Selecting columns

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
```

Select one column by placing its name in square brackets.

```python
cities["Name"]
```

Select several columns by supplying a list of column names. The double brackets appear because the inner brackets create the list.

```python
cities[["Name", "Population, 2021"]]
```

The original DataFrame is unchanged. Store the result in a new variable when you want to reuse it.

```python
city_populations = cities[["Name", "Population, 2021"]]
city_populations.head()
```
