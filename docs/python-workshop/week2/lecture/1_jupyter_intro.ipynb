# Jupyter notebooks

A Jupyter notebook is divided into **cells**. Code cells run Python, while Markdown cells hold headings, instructions, and explanations.

Run the code cells from top to bottom. A later cell can use variables created by an earlier cell.

```python
city = "Toronto"
population = 2_794_356

print(city)
print(population)
```

After running the cell above, the variables remain available to later cells. Editing a cell does not change its output until you run it again.

```python
print(city, "has a population of", population)
```

If results become confusing because cells were run out of order, use **Restart Kernel and Run All**. This clears stored values and reruns the notebook from the beginning.
