# Filtering rows

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
```

A comparison applied to a column produces `True` or `False` for every row.

```python
cities["Population, 2021"] >= 100_000
```

Use `.loc[]` to retain the rows where the comparison is true.

```python
large_cities = cities.loc[
    cities["Population, 2021"] >= 100_000
]

large_cities.head()
```

Text columns can be filtered in the same way.

```python
ontario_cities = cities.loc[
    cities["Prov/terr"] == "Ont."
]

ontario_cities.head()
```
