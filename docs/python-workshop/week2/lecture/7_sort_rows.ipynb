# Sorting rows

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
```

Sort from largest to smallest by setting `ascending=False`.

```python
largest_first = cities.sort_values(
    by="Population, 2021",
    ascending=False
)

largest_first.head(10)
```

Filtering and sorting can be completed in separate, easy-to-inspect steps.

```python
ontario_cities = cities.loc[
    cities["Prov/terr"] == "Ont."
]

ontario_cities = ontario_cities.sort_values(
    by="Population, 2021",
    ascending=False
)

ontario_cities.head(10)
```
