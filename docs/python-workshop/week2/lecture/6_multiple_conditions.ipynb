# Filtering with multiple conditions

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
```

Use `&` for **and**. Put each pandas condition inside parentheses.

```python
large_ontario_cities = cities.loc[
    (cities["Prov/terr"] == "Ont.")
    & (cities["Population, 2021"] >= 100_000)
]

large_ontario_cities
```

Use `|` for **or**.

```python
ontario_or_quebec = cities.loc[
    (cities["Prov/terr"] == "Ont.")
    | (cities["Prov/terr"] == "Que.")
]

ontario_or_quebec.head()
```

These operators play the same role as `and` and `or` from Week 1, but pandas uses `&` and `|` when comparing whole columns.
