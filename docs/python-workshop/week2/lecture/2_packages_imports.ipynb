# Packages and imports

Python includes built-in libraries such as `csv`. Other packages, such as pandas, add tools that are installed separately.

Pandas should already be installed from Module 0. Installing usually happens once; importing happens whenever a notebook uses the package.

From a terminal, pandas can be installed with:

```bash
python -m pip install pandas
```

On some macOS or Ubuntu computers, the command may begin with `python3` instead.

````python
import pandas as pd

print("pandas version:", pd.__version__)
````

`pd` is the conventional short name for pandas. It keeps later commands shorter, such as `pd.read_csv(...)`.
