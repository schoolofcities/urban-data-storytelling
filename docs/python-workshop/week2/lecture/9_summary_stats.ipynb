# Summary statistics and counts

```python
import pandas as pd

cities = pd.read_csv("../data/cities.csv")
cities = cities.rename(
    columns={
        "Name": "city",
        "Prov/terr": "province",
        "Population, 2021": "population_2021",
        "Population, 2016": "population_2016"
    }
)
```

```python
cities["population_2021"].count()
```

```python
cities["population_2021"].mean()
```

```python
cities["population_2021"].median()
```

```python
cities["population_2021"].min()
```

```python
cities["population_2021"].max()
```

```python
cities["population_2021"].describe()
```

Use `value_counts()` to count the number of rows in each category.

```python
cities["province"].value_counts()
```
