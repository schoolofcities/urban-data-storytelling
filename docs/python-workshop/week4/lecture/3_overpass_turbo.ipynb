# 3. Seeing the query directly with Overpass Turbo

OSMnx makes OpenStreetMap much easier to use from Python, but underneath it relies on OpenStreetMap data-access services.

[Overpass Turbo](https://overpass-turbo.eu/) is a web interface where you can run an OpenStreetMap query and see the results immediately on a map.

It is useful for understanding what a query is actually asking for.

## Try the Wizard first

In Overpass Turbo:

1. move the map to Toronto;
2. click **Wizard**;
3. enter:

```text
amenity=library
```

4. run the query.

The result should highlight mapped libraries in the visible area.

## Example Overpass query

The Wizard generates code similar in spirit to this:

```text
[out:json][timeout:25];

nwr["amenity"="library"]({{bbox}});

out geom;
```

At this stage, the important pieces are only:

| Part | Intuition |
|---|---|
| `[out:json]` | return structured JSON data |
| `nwr` | search nodes, ways, and relations |
| `["amenity"="library"]` | keep features tagged as libraries |
| `{{bbox}}` | search the map area currently visible |
| `out geom` | return geometry |

## Change one thing at a time

Try replacing the tag with:

```text
amenity=school
```

or:

```text
leisure=park
```

or:

```text
amenity=bicycle_parking
```

Notice how the query structure stays almost identical while the requested feature changes.

## Why are we not learning Overpass QL in depth?

Overpass is powerful, but its query language can become complicated quickly.

For routine urban analysis, it is completely reasonable to:

- adapt an existing query;
- use the Overpass Turbo Wizard;
- consult documentation;
- use an AI coding tool to help construct a query;
- use OSMnx instead when it provides the feature access you need.

The important skill is being able to **read the query well enough to understand what is being requested and verify the result**.
