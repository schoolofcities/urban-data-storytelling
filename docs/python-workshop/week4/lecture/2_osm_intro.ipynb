# 2. Exploring OpenStreetMap

[OpenStreetMap](https://www.openstreetmap.org/) is a collaborative geographic database of features on the ground.

It can contain:

- roads and paths;
- buildings;
- parks;
- transit infrastructure;
- libraries and schools;
- shops and services;
- many other mapped urban features.

OpenStreetMap is not a census. It is better thought of as a large, continuously edited database of geographic features.

## Inspect an actual feature

Open this search in a browser:

[Search OpenStreetMap for Toronto Reference Library](https://www.openstreetmap.org/search?query=Toronto%20Reference%20Library)

Click the relevant result and inspect the feature information.

Look for fields such as:

- `name`
- `amenity`
- `operator`
- `addr:*`
- opening hours or contact information, when present

The exact set of fields differs from feature to feature.

## Tags: how OpenStreetMap describes features

OpenStreetMap commonly describes features using **key/value tags**.

| Feature | Example tag |
|---|---|
| Library | `amenity=library` |
| School | `amenity=school` |
| Park | `leisure=park` |
| Bicycle parking | `amenity=bicycle_parking` |
| Railway station | `railway=station` |
| Supermarket | `shop=supermarket` |

These tags are the basic idea behind the queries we will make in Python.

## Points, lines, and polygons still matter

The same kind of feature can sometimes appear with different geometry types.

For example, a library might be mapped as:

- a **point** marking its location; or
- a **polygon** representing the library building.

This is why we always inspect `geometry.geom_type` after downloading OSM data.

## A useful habit

When you want to query a new feature type:

1. inspect OpenStreetMap itself;
2. identify likely tags;
3. test the tags in Overpass Turbo or OSMnx;
4. inspect the returned records;
5. revise the query if the result does not represent what you intended.

You do not need to memorize the OpenStreetMap tagging system.
