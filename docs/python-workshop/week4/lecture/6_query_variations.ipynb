# 6. Adapting the OSMnx query

The usefulness of OSMnx is not specific to libraries.

The same function can retrieve many types of urban features by changing the `tags` dictionary.

````python
import osmnx as ox

place = "Toronto, Ontario, Canada"
````

## Example 1: schools

```python
tags = {"amenity": "school"}

schools = ox.features_from_place(
    place,
    tags=tags
)
```

## Example 2: parks

```python
tags = {"leisure": "park"}

parks = ox.features_from_place(
    place,
    tags=tags
)
```

## Example 3: bicycle parking

```python
tags = {"amenity": "bicycle_parking"}

bicycle_parking = ox.features_from_place(
    place,
    tags=tags
)
```

## Example 4: railway stations

```python
tags = {"railway": "station"}

stations = ox.features_from_place(
    place,
    tags=tags
)
```

## Several values for the same key

OSMnx also allows a list of acceptable values:

```python
tags = {
    "amenity": ["library", "school"]
}
```

This means:

> return features where `amenity` is library **or** school.

Always inspect the output to confirm that the query produced the feature set you intended.

## One caution about large queries

A query such as:

```python
{"building": True}
```

could request an enormous number of features across Toronto.

Start with a focused tag or a smaller geographic area, inspect the result, and expand only when necessary.
