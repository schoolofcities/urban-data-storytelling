## A typical urban-data workflow

```text
Urban question
    ↓
Find a data source
    ↓
Retrieve or load the data
    ↓
Inspect the raw data
    ↓
Clean and transform it
    ↓
Combine it with other data
    ↓
Analyse or map it
    ↓
Save useful outputs
    ↓
Document what you did
```

A workflow should be understandable not only to Python, but also to another person (including your future self).

## Example question for today

> How are public libraries distributed across Toronto census areas?

We will combine:

1. **library locations from OpenStreetMap**, retrieved with OSMnx;
2. **Toronto census polygons**, already stored in the course data.

Once the workflow works for libraries, the same structure can be adapted to other urban features such as parks, schools, pharmacies, bicycle parking, or rail stations.

## Why save intermediate data?

API and OpenStreetMap queries depend on external services.

Rather than repeatedly downloading the same data, a good workflow often:

```text
download once
    ↓
save raw result
    ↓
reopen saved file
    ↓
clean and analyse locally
```

This is also why every lecture notebook this week is **self-contained**: later notebooks read saved files instead of assuming that variables created in another notebook still exist.
