# 10. Organizing a reproducible workflow

A small project might look like:

```text
week4/
├── assignment/
├── data/
│   ├── raw/
│   ├── processed/
│   └── practice/
├── exercises/
├── lecture/
├── output/
└── reference/
```

The exact structure is flexible. The important idea is to separate **source data**, **processed data**, **analysis code**, and **final outputs**.

## Raw versus processed

**Raw data**

- downloaded or supplied source data;
- should generally remain unchanged;
- allows the workflow to be rerun.

**Processed data**

- cleaned columns;
- transformed geometry;
- joined datasets;
- calculated variables.

**Outputs**

- final tables;
- maps;
- figures;
- report-ready files.

## Final checklist

A reproducible workflow should:

1. import packages near the beginning;
2. define important settings clearly;
3. use relative file paths;
4. create required folders in code when appropriate;
5. inspect data before transforming it;
6. save expensive or externally retrieved data;
7. check joins and calculated variables;
8. explain important decisions in Markdown;
9. run from top to bottom;
10. document sources and retrieval dates.
