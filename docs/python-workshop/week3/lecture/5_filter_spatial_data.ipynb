# 5. Filtering spatial data

A GeoDataFrame can be filtered using the same pandas syntax from Week 2.

The key difference is that every retained row keeps its geometry, so the filtered result can immediately be mapped.

```python
import geopandas as gpd

stops = gpd.read_file("../data/ttc_stops.geojson")
routes = gpd.read_file("../data/ttc_routes.geojson")
```

## Inspect categories before filtering

Do not guess the exact text stored in a column. First inspect the values.

```python
print(stops["NAME"].value_counts())
```

## Filter one transit line

This is ordinary pandas Boolean filtering.

```python
line_5_stops = stops.loc[
    stops["NAME"] == "Line 5: Eglinton Crosstown LRT"
]

line_5_stops[["LOCATION_N", "NAME", "STATUS"]].head()
```

```python
print("All stops:", len(stops))
print("Line 5 stops:", len(line_5_stops))
```

Because `line_5_stops` is still a GeoDataFrame, we can map the subset immediately.

```python
line_5_stops.plot(figsize=(8, 4), markersize=25)
```

## Filter using another attribute

For example, keep only routes whose status is `Existing`.

```python
existing_routes = routes.loc[
    routes["STATUS"] == "Existing"
]

existing_routes[["NAME", "STATUS"]]
```

## Combine conditions

The syntax is the same as pandas:

- `&` means **and**
- `|` means **or**
- put each condition in parentheses

```python
existing_line_5_stops = stops.loc[
    (stops["NAME"] == "Line 5: Eglinton Crosstown LRT")
    & (stops["STATUS"] == "Existing")
]

print("Matching stops:", len(existing_line_5_stops))
```

Filtering before mapping or joining is often useful because it lets the analysis focus only on the features relevant to the question.
