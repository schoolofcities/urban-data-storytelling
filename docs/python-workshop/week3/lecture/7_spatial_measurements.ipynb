# 7. From geometry to spatial measurements

Geometry is not only for drawing maps. GeoPandas can calculate properties of geographic features.

For polygons, useful properties include:

- area;
- centroid;
- bounds.

For lines, useful properties include length.

The most important rule is that the result uses the units of the CRS.

```python
import geopandas as gpd

wards = gpd.read_file("../data/city-wards.gpkg")

print("Ward CRS:", wards.crs)
```

Because the ward CRS is measured in metres:

- polygon `.area` is in square metres;
- line `.length` would be in metres.

```python
print("First ward area in square metres:")
print(wards.geometry.area.iloc[0])
```

A centroid is a representative centre point of a polygon. It can be useful for labels or rough visualization, although it is not always the best measure of accessibility or a true population centre.

```python
ward_centres = wards.copy()
ward_centres["geometry"] = ward_centres.geometry.centroid

ward_centres[["WARD_NAME", "geometry"]].head()
```
