# 3. Loading and inspecting GeoDataFrames

Most pandas inspection methods from Week 2 work unchanged in GeoPandas.

Before mapping or calculating anything, inspect the data first.

````python
import geopandas as gpd

stops = gpd.read_file("../data/ttc_stops.geojson")
````

## Familiar pandas checks

Start with the table itself.

````python
stops.head()
````

````python
print("Rows and columns:", stops.shape)
print("Columns:")
print(stops.columns.tolist())
````

We can also inspect categorical values before deciding what to filter or map.

````python
print(stops["NAME"].value_counts())
print("Status values:")
print(stops["STATUS"].value_counts())
````

## Spatial checks

GeoPandas adds several properties that are especially useful when first opening a spatial file:

- `.geometry.geom_type` — geometry type for each feature
- `.crs` — coordinate reference system
- `.total_bounds` — minimum and maximum coordinates for the entire layer
- `.is_valid` — whether each geometry is structurally valid

````python
print("Geometry types:")
print(stops.geometry.geom_type.value_counts())

print("CRS:", stops.crs)
print("Bounds:", stops.total_bounds)
print("Invalid geometries:", (~stops.geometry.is_valid).sum())
````

`total_bounds` returns:

```text
[min_x, min_y, max_x, max_y]
```

For longitude/latitude data, those values give the west, south, east, and north extent of the dataset.

````python
min_x, min_y, max_x, max_y = stops.total_bounds

print("West:", min_x)
print("South:", min_y)
print("East:", max_x)
print("North:", max_y)
````

A useful habit is to answer three questions before proceeding:

1. **What does one row represent?**
2. **What type of geometry is present?**
3. **What CRS is the geometry using?**
