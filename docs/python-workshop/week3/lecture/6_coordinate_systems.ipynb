# 6. Coordinate reference systems and projections

A coordinate reference system, or **CRS**, explains how numeric coordinates correspond to locations on Earth.

A map can look reasonable while still using a CRS that is inappropriate for the calculation you want to perform.

```python
import geopandas as gpd

stops = gpd.read_file("../data/ttc_stops.geojson")
wards = gpd.read_file("../data/city-wards.gpkg")

print("Stops CRS:", stops.crs)
print("Wards CRS:", wards.crs)
```

## Two useful examples

### EPSG:4326

- longitude and latitude;
- coordinates measured in **degrees**;
- common for web data and GPS locations;
- good for storing locations across the world;
- not appropriate for directly measuring local area or distance.

### EPSG:32617

- UTM Zone 17N;
- coordinates measured in **metres**;
- suitable for local measurements around Toronto.

## Coordinates look different after projection

The location does not move in the real world. Only the numeric coordinate system changes.

```python
print("Original stop geometry:")
print(stops.geometry.iloc[0])

stops_projected = stops.to_crs(wards.crs)

print("Projected stop geometry:")
print(stops_projected.geometry.iloc[0])
```

The first pair resembles longitude and latitude. The projected version contains much larger values because its coordinates are measured in metres within the UTM coordinate system.

## Reproject with `.to_crs()`

Use `.to_crs()` when the existing CRS is known and you want to **convert the coordinates**.

```python
stops_projected = stops.to_crs(wards.crs)

print("Original CRS:", stops.crs)
print("Projected CRS:", stops_projected.crs)
```

## `.to_crs()` versus `.set_crs()`

These are not interchangeable:

| Method | What it does |
|---|---|
| `.to_crs()` | converts geometry coordinates into another CRS |
| `.set_crs()` | labels existing coordinates with a CRS without moving/converting them |

For ordinary reprojection in this course, use **`.to_crs()`**.

Using `set_crs()` to force mismatched layers to have the same label can create incorrect geography.

## Practical rule for this course

- For **displaying longitude/latitude locations**, EPSG:4326 is often fine.
- For **area, distance, and buffers around Toronto**, use an appropriate projected CRS such as the supplied ward CRS.
- Before combining layers, check `.crs` rather than assuming they match.
