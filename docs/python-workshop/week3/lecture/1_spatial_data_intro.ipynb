# 1. Spatial data and urban questions

Spatial data records **where** something is, as well as information about it.

A normal pandas table might tell us:

| facility | type | year_opened |
|---|---|---:|
| Fort York Library | Library | 2014 |
| Wychwood Library | Library | 1916 |

Spatial data adds information describing **location or shape**. That allows us to ask questions such as:

- Where are transit stops and routes?
- Which ward contains the most libraries?
- How large is each ward?
- Which facilities are close to a transit line?
- How are services distributed across Toronto?

GeoPandas extends pandas so that we can work with this geographic information using familiar DataFrame operations.

```python
import geopandas as gpd
```

## Three common vector geometry types

| Geometry | What it represents | Urban examples |
|---|---|---|
| **Point** | one location | transit stop, collision, library |
| **Line** | a path | transit route, street, bike lane |
| **Polygon** | an area | ward, park, parcel, neighbourhood |

A spatial dataset normally uses one primary geometry column to describe each row.

```python
stops = gpd.read_file("../data/ttc_stops.geojson")
routes = gpd.read_file("../data/ttc_routes.geojson")
wards = gpd.read_file("../data/city-wards.gpkg")

print("Stops:", stops.geometry.geom_type.unique())
print("Routes:", routes.geometry.geom_type.unique())
print("Wards:", wards.geometry.geom_type.unique())
```

## One row still represents one feature

GeoPandas does not replace the tabular model from Week 2.

For example, one row in the TTC stop data represents one station or stop. Its ordinary columns describe the feature, while `geometry` stores its location.

```python
stops[["LOCATION_N", "NAME", "STATUS", "geometry"]].head()
```

This is the key idea for the week:

> **A GeoDataFrame is a DataFrame with geographic geometry attached to each row.**
