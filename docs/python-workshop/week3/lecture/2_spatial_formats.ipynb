# 2. Common spatial data formats

Urban spatial data is distributed in several formats. GeoPandas can read many of them with the same function: `gpd.read_file()`.

## Formats you are likely to encounter

| Format | Typical use | Notes |
|---|---|---|
| **GeoJSON** | web/open-data distribution | one readable text-based file |
| **GeoPackage (`.gpkg`)** | modern GIS storage | one file; can contain multiple layers |
| **Shapefile** | older GIS datasets | several related files must stay together |
| **CSV** | tabular data | may contain coordinates but usually has no geometry column yet |

For this course, GeoJSON and GeoPackage are the most convenient formats.

```python
import geopandas as gpd

stops = gpd.read_file("../data/ttc_stops.geojson")
wards = gpd.read_file("../data/city-wards.gpkg")

print(type(stops))
print(type(wards))
```

Both files become the same Python object type: a `GeoDataFrame`.

That means the file format matters when storing and sharing data, but much of the analysis code remains the same after the file is loaded.

```python
print("Stop columns:")
print(stops.columns.tolist())

print("Ward columns:")
print(wards.columns.tolist())
```

## The geometry column

GeoPandas keeps one **active geometry column**. By default it is usually named `geometry`.

```python
print("Active geometry column:", stops.geometry.name)
print(stops.geometry.head())
```

Geometry values may look unfamiliar when printed as text (`POINT`, `LINESTRING`, `POLYGON`), but GeoPandas understands them as geographic objects that can be mapped and measured.

## A note on shapefiles

A shapefile is not actually one file. A dataset commonly includes `.shp`, `.shx`, `.dbf`, and `.prj` files. If one is missing, the dataset may not work correctly.

GeoPackage avoids this problem by storing the spatial information in one `.gpkg` file.
