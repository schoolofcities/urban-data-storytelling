# 4. Creating maps and combining spatial layers

GeoPandas `.plot()` creates quick static maps. These are especially useful for inspecting spatial data before doing more detailed analysis.

````python
import geopandas as gpd
import matplotlib.pyplot as plt

stops = gpd.read_file("../data/ttc_stops.geojson")
routes = gpd.read_file("../data/ttc_routes.geojson")
wards = gpd.read_file("../data/city-wards.gpkg")
````

## Plot one layer

A single call is enough for a basic exploratory map.

````python
stops.plot()
````

Useful arguments include:

- `figsize` — map dimensions
- `markersize` — point size
- `linewidth` — line thickness
- `edgecolor` — polygon outlines

We will keep styling minimal and focus on the data.

````python
routes.plot(figsize=(8, 6), linewidth=2)
````

## Before combining layers: check CRS

Layers need to use compatible coordinate systems before they can be drawn together correctly.

````python
print("Stops CRS:", stops.crs)
print("Routes CRS:", routes.crs)
print("Wards CRS:", wards.crs)
````

The TTC files and ward file use different CRS values. Reproject the TTC layers to the ward CRS before combining them.

````python
stops_projected = stops.to_crs(wards.crs)
routes_projected = routes.to_crs(wards.crs)
````

## Combine layers on one Matplotlib axis

The basic pattern is:

```text
create an axis
→ draw polygons
→ draw lines
→ draw points
```

Each layer receives the same `ax=ax` argument.

````python
fig, ax = plt.subplots(figsize=(8, 8))

wards.plot(
    ax=ax,
    facecolor="none",
    edgecolor="black",
    linewidth=0.6
)

routes_projected.plot(
    ax=ax,
    linewidth=2
)

stops_projected.plot(
    ax=ax,
    markersize=8
)

ax.set_title("Toronto rapid-transit routes and stops")
ax.set_axis_off()
````

## Layer order matters

If a filled polygon layer is drawn last, it can cover the lines and points underneath it.

A common order is therefore:

1. background polygons;
2. lines;
3. points.

The same layering idea will be useful later when combining boundaries with facilities, routes, or other urban features.
